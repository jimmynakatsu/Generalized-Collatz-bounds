```python
# 2. Render the Collatz trajectory animation
%%manim -qm -v WARNING CollatzTrajectory

from manim import *

class CollatzTrajectory(Scene):
    def construct(self):
        n = 27
        seq = [n]
        while n != 1:
            n = (3 * n + 1) if (n % 2 != 0) else (n // 2)
            seq.append(n)

        axes = Axes(
            x_range=[0, len(seq) + 5, 20],
            y_range=[0, 10000, 2000],
            x_length=10,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 20},
        ).shift(DOWN * 0.5)

        labels = axes.get_axis_labels(
            x_label=Text("Steps", font_size=22),
            y_label=Text("Value", font_size=22)
        )
        title = Title("Collatz Trajectory: n = 27 (Peak: 9,232)", font_size=32)

        self.play(Write(title), Create(axes), Write(labels))

        dot = Dot(color=YELLOW).move_to(axes.c2p(0, seq[0]))
        trace = VMobject(color=BLUE, stroke_width=3)
        trace.set_points_as_corners([axes.c2p(0, seq[0])])

        self.add(trace, dot)

        for i in range(1, len(seq)):
            next_pt = axes.c2p(i, seq[i])
            run_time = 0.08 if i < 40 else 0.03
            self.play(
                dot.animate.move_to(next_pt),
                trace.animate.add_points_as_corners([next_pt]),
                run_time=run_time,
                rate_func=linear
            )
        self.wait(2)
```
